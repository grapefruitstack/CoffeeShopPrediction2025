### How did I clean the data?

Cleaning the data consisted of removing an overlapping month (March). The original data documented March 2024 AND March 2025, which skewed the data to anything that was favored in March. Therefore, I removed the March of 2025 and kept the March of 2024.

I also added missing dates to the data in the dates column, which was easy because there's already the listed month and day. This would be especially useful in heatmaps by date.

# Coffee Sales By Weekday

```python
# @title
# Uncleaned Coffee Sales by Weekday

drink_counts = CoUn.groupby(['Weekdaysort', 'coffee_name']).size().unstack(fill_value=0)
drink_counts = drink_counts.sort_index()

drink_counts.plot(kind='bar', figsize=(10,6))
plt.title("Uncleaned Coffee Sales by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Number of Orders")
plt.legend(title="Coffee Type")
plt.show()
```

### Unclean vs Clean for Weekday

*   The max amount of orders has decreased.
*   There is less distance between Cappucinno and Americano.



```python
# @title
# Cleaned Coffee Sales by Weekday

drink_counts = CoCl.groupby(['Weekdaysort', 'coffee_name']).size().unstack(fill_value=0)
drink_counts = drink_counts.sort_index()

drink_counts.plot(kind='bar', figsize=(10,6))

plt.title("Coffee Sales by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Number of Orders")
plt.legend(title="Coffee Type")
plt.show()
```

## Coffee Sales By Time of Day

```python
# @title
melted_clean = CoCl.groupby(['Time_of_Day', 'coffee_name']).size().reset_index(name='sales')
melted_unclean = CoUn.groupby(['Time_of_Day', 'coffee_name']).size().reset_index(name='sales')

time_order = ['Morning', 'Afternoon', 'Night']

# Convert Time_of_Day to a categorical type with an explicit order
melted_unclean['Time_of_Day'] = pd.Categorical(
    melted_unclean['Time_of_Day'],
    categories=time_order,
    ordered=True
)

# Then plot
sns.lineplot(
    data=melted_unclean,
    x='Time_of_Day',
    y='sales',
    hue='coffee_name',
    marker='o'
)
plt.title("Uncleaned Coffee Sales by Time of Day")
plt.xlabel("Time of Day")
plt.ylabel("Number of Orders")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title="Coffee Type")
plt.tight_layout()
plt.show()
```

### Unclean vs Clean for Time of Day

*   March data seemed to have dramatically favored Americano, seeing as most of its stats are higher in the uncleaned data.


```python
# @title
time_order = ['Morning', 'Afternoon', 'Night']

melted_clean['Time_of_Day'] = pd.Categorical(
    melted_clean['Time_of_Day'],
    categories=time_order,
    ordered=True
)

sns.lineplot(
    data=melted_clean,
    x='Time_of_Day',
    y='sales',
    hue='coffee_name',
    marker='o'
)
plt.title("Coffee Sales by Time of Day")
plt.xlabel("Time of Day")
plt.ylabel("Number of Orders")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title="Coffee Type")
plt.tight_layout()
plt.show()
```

## Outliers

```python
# @title
outliers_list = []

for name, group in CoUn.groupby('coffee_name'):
    daily = group.groupby('Date').size()
    Q1 = daily.quantile(0.25)
    Q3 = daily.quantile(0.75)
    IQR = Q3 - Q1
    mask = (daily < Q1 - 3 * IQR) | (daily > Q3 + 3 * IQR)
    outlier_days = daily[mask]
    if not outlier_days.empty:
        outliers_list.append(pd.DataFrame({
            'coffee_name': name,
            'Date': outlier_days.index,
            'sales': outlier_days.values
        }))

outliers_df = pd.concat(outliers_list)
pd.set_option('display.max_rows', None)
print(outliers_df)
```
